## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a classification problem — I'm trying to sort pages into "declining" or "not declining." But the actual output I want isn't just a yes/no, it's a ranked list, so I use the model's probability score to sort pages from most-likely-to-need-review to least.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

I'm using trend_direction == "down" as my target for now. It's not a perfect target though — it's just a label based on where things stand right now, not whether the page will actually keep declining. A better version later would predict what happens in the next 30 days based on the last 90 days, instead of just labeling the current snapshot.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I'm using Precision@50. The reason is simple — a reviewer can only realistically check about 50 pages, not all 30,000. So what matters isn't "is the model right overall," it's "are the top 50 pages it picks actually worth checking." Regular accuracy wouldn't mean much here since most pages are boring and easy to get right by default.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
baseline_precision_at_50 = 0.240
model_precision_at_50 = 0.740
print(f"Baseline rule: {baseline_precision_at_50}  |  Learned model: {model_precision_at_50}")
```

A simple rule breaks down because "declining" depends on more than one thing at once. A stale page with barely any traffic was never worth reviewing anyway, but a stale page that used to get a lot of traffic is a real problem — same "staleness," totally different situation depending on the traffic. A basic if-statement can't easily juggle that combo. I already tested this myself: my hand-written rule only got 24% right on its top 50 picks, while a simple decision tree got 82% — proof the real pattern isn't a single clean threshold.
