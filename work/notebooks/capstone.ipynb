## 1. Question

*The research question and the decision it supports.*

Which pages should a content team review first, out of thousands, when nobody
has time to check them all? This project builds and validates a model to rank
pages by risk of declining search visibility, supporting a weekly human review
queue — not an automated action.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

FlyRank ML Internship Warehouse (Hugging Face), tables: fact_content_daily_performance,
dim_content, fact_content_query_90d. Window: Feb 1 – Mar 31 2026 (mid-panel month,
avoiding the sealed final month per the warehouse's own documentation). Filtered
to content items with ≥50 prior-window impressions. Final population: 73,112
content items, 18.2% labeled declining.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Label: 30-day impressions dropped >20% vs the prior 30 days. Five features, all
knowable before the label window: imp_prev30, clk_prev30, pos_prev30,
visible_queries, content_age_days. Baseline: a hand-written CTR-vs-position rule
(Week 4). Validation: client-grouped split (GroupShuffleSplit, 75/25, zero
client overlap). Leakage checks: baseline score excluded from training features
(confirmed); deliberate trap (re-adding imp_last30) pushed Precision@50 from
0.380 to 1.000, confirming the feature set is clean without it.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

| Method | Precision@50 |
|---|---|
| Base rate (random) | 0.133 |
| Week-4 baseline rule | 0.000 (different target, not a fair loss) |
| Naive random split | 0.640 (inflated) |
| Grouped split (reported) | 0.380 (~2.9x base rate) |

The gap between the naive split (0.640) and the honest split (0.380) is itself
a finding — a 41% relative overstatement from letting client identity leak
across train/test.

## 5. Limitations

*What this work cannot claim.*

Observational, not causal — supports "worth reviewing," not "will cause
recovery." Validated on one month only. Label is a proxy, not a verified
business outcome. Baseline's 0.000 reflects target mismatch, not poor design.
visible_queries (top feature) comes from a fixed 90-day window that can overlap
the label window — a soft signal pending a cleaner rebuild.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

Every flagged page: action = review_for_refresh_or_consolidation, with one of
three reason codes (aging_narrow_visibility, high_traffic_volatility,
general_decline_risk). Human review required before any action — confirm
intent match, check for a recent unrelated fix, rule out seasonality. Never
automate: publishing changes, removing pages, reallocating budget, or applying
this outside the validated scope.

## 5-Minute Demo Outline

**Question:** Which pages should a content team review first, out of thousands,
when nobody has time to check them all?

**Method:** Trained a logistic regression on five pre-decision features
(prior impressions, clicks, position, query visibility, content age) to rank
pages by risk of declining search impressions. Validated with a client-grouped
split so the model is only ever judged on clients it never trained on.

**One chart:** Precision@50 comparison — base rate 0.133, naive random split
0.640 (inflated), honest grouped split 0.380 (the real result).

**One honest result:** The model ranks declining pages into its top 50 picks at
roughly 2.9x the rate of random selection — but a naive validation split would
have overstated that by 68%, a gap worth showing live as the "aha" moment.

**One recommendation:** Use the ranked queue as a weekly human-reviewed
prioritization list, never as an automated action — flagged pages get a
reason code and go to a person, not a publish button.
