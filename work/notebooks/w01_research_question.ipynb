Action: a reviewer opens the top-ranked pages and decides whether to refresh, expand, or leave each one alone.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

With 30,000 pages and 54.2% of them currently trending down, there are far more candidate pages than a reviewer can manually check by hand. Search volume barely correlates with actual impressions (0.001), meaning a naive "high search volume = worth reviewing" rule wouldn't capture the right pages — which is exactly the kind of gap a ranked model can close instead.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/samhere3116-maker/ML-Pipeline/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].drop_duplicates(subset="content_id")

print(f"Rows in starter dataset after basic filters: {len(df)}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")

declining_share = (df["trend_direction"] == "down").mean()
print(f"Share of pages currently trending down: {declining_share:.1%}")
```
